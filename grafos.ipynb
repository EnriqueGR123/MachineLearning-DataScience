```python
import networkx as nx
import matplotlib.pyplot as plt


# -----------------------------------------
# DATOS DE LAS RUTAS
# -----------------------------------------

rutas = {
    "Ruta 1 - Autopista": {
        "nodos": ["Guadalajara", "El Arenal", "Plan de Barrancas",
                  "Amado Nervo", "Bucerías", "Puerto Vallarta"],
        "casetas": [205, 306, 205, 0, 664]
    },

    "Ruta 2 - Libre": {
        "nodos": ["Guadalajara", "Tala", "Ameca", "Mixtlán",
                  "Mascota", "Las Palmas", "Puerto Vallarta"],
        "casetas": [0, 0, 0, 0, 0, 0]
    },

    "Ruta 3 - Mixta": {
        "nodos": ["Guadalajara", "El Arenal", "Plan de Barrancas",
                  "Bucerías", "Puerto Vallarta"],
        "casetas": [205, 306, 0, 664]
    }
}


# -----------------------------------------
# SOLICITAR DATOS AL USUARIO
# -----------------------------------------

origen = input("Ingresa el punto de partida: ")
destino = input("Ingresa el destino: ")

print("\n======================================")
print("          RESULTADOS DEL VIAJE")
print("======================================")

print("Origen:", origen)
print("Destino:", destino)


# -----------------------------------------
# MOSTRAR LAS RUTAS
# -----------------------------------------

for nombre, ruta in rutas.items():

    print("\n--------------------------------------")
    print(nombre)
    print("--------------------------------------")

    nodos = ruta["nodos"]
    casetas = ruta["casetas"]

    # Cambiamos el primer y último nodo
    nodos[0] = origen
    nodos[-1] = destino

    print("Ruta:")

    for i in range(len(nodos) - 1):
        print(f"{nodos[i]} -> {nodos[i + 1]}")

    # Calcular costo total
    costo_total = sum(casetas)

    print("Costo de peaje: $", costo_total, "MXN")


# -----------------------------------------
# DIBUJAR LOS GRAFOS
# -----------------------------------------

for nombre, ruta in rutas.items():

    G = nx.Graph()

    nodos = ruta["nodos"]

    # Agregar nodos
    for nodo in nodos:
        G.add_node(nodo)

    # Agregar conexiones
    for i in range(len(nodos) - 1):
        G.add_edge(nodos[i], nodos[i + 1])

    # Posiciones
    posiciones = nx.spring_layout(G, seed=10)

    plt.figure(figsize=(10, 5))

    nx.draw(
        G,
        posiciones,
        with_labels=True,
        node_color="lightblue",
        node_size=2500,
        font_size=9,
        font_weight="bold",
        edge_color="gray"
    )

    plt.title(nombre)

    plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt


# ==========================================================
# CREAR EL GRAFO
# ==========================================================

G = nx.Graph()

# Cada conexión tiene:
# - peaje = costo de casetas
# - tiempo = tiempo estimado en minutos

G.add_edge("Guadalajara", "El Arenal", peaje=205, tiempo=40)
G.add_edge("Guadalajara", "Tala", peaje=0, tiempo=50)

G.add_edge("El Arenal", "Plan de Barrancas", peaje=306, tiempo=75)
G.add_edge("El Arenal", "Ameca", peaje=0, tiempo=90)

G.add_edge("Plan de Barrancas", "Amado Nervo", peaje=205, tiempo=50)
G.add_edge("Plan de Barrancas", "Mascota", peaje=0, tiempo=150)

G.add_edge("Amado Nervo", "Bucerías", peaje=0, tiempo=60)
G.add_edge("Amado Nervo", "Compostela", peaje=0, tiempo=45)

G.add_edge("Bucerías", "Puerto Vallarta", peaje=664, tiempo=40)
G.add_edge("Bucerías", "Las Palmas", peaje=0, tiempo=70)

G.add_edge("Compostela", "Bucerías", peaje=439, tiempo=80)
G.add_edge("Compostela", "Puerto Vallarta", peaje=0, tiempo=180)

G.add_edge("Tala", "Ameca", peaje=0, tiempo=80)
G.add_edge("Ameca", "Mixtlán", peaje=0, tiempo=100)

G.add_edge("Mixtlán", "Mascota", peaje=0, tiempo=100)
G.add_edge("Mascota", "Las Palmas", peaje=0, tiempo=120)

G.add_edge("Las Palmas", "Puerto Vallarta", peaje=0, tiempo=60)


# ==========================================================
# FUNCIÓN PARA MOSTRAR UNA RUTA
# ==========================================================

def mostrar_ruta(nombre, ruta, criterio):

    print("\n" + "=" * 60)
    print(nombre)
    print("=" * 60)

    print("\nRecorrido:")

    costo_total = 0
    tiempo_total = 0

    for i in range(len(ruta) - 1):

        origen = ruta[i]
        destino = ruta[i + 1]

        datos = G[origen][destino]

        costo_total += datos["peaje"]
        tiempo_total += datos["tiempo"]

        print(
            f"{origen} -> {destino} "
            f"(Peaje: ${datos['peaje']} | "
            f"Tiempo: {datos['tiempo']} min)"
        )

    print("\nCosto total de peajes: $" + str(costo_total) + " MXN")
    print("Tiempo estimado:", tiempo_total, "minutos")

    horas = tiempo_total // 60
    minutos = tiempo_total % 60

    print(f"Tiempo aproximado: {horas} h {minutos} min")


# ==========================================================
# DIBUJAR EL GRAFO COMPLETO
# ==========================================================

def dibujar_grafo(ruta1, ruta2, ruta3):

    posiciones = nx.spring_layout(G, seed=20)

    plt.figure(figsize=(14, 9))

    # Dibujar todos los nodos y conexiones
    nx.draw(
        G,
        posiciones,
        with_labels=True,
        node_color="lightgray",
        node_size=2500,
        font_size=9,
        font_weight="bold",
        edge_color="gray"
    )

    # Dibujar las rutas encontradas con diferentes colores

    def dibujar_ruta(ruta, color):

        conexiones = []

        for i in range(len(ruta) - 1):
            conexiones.append((ruta[i], ruta[i + 1]))

        nx.draw_networkx_edges(
            G,
            posiciones,
            edgelist=conexiones,
            edge_color=color,
            width=4
        )

    dibujar_ruta(ruta1, "green")
    dibujar_ruta(ruta2, "blue")
    dibujar_ruta(ruta3, "orange")

    plt.title(
        "Rutas entre origen y destino",
        fontsize=16,
        fontweight="bold"
    )

    plt.show()


# ==========================================================
# PROGRAMA PRINCIPAL
# ==========================================================

print("=" * 60)
print("       SISTEMA DE RUTAS CON GRAFOS Y DIJKSTRA")
print("=" * 60)

print("\nLugares disponibles:")

lugares = list(G.nodes)

for lugar in lugares:
    print("-", lugar)


# ==========================================================
# PEDIR ORIGEN Y DESTINO
# ==========================================================

origen = input("\nIngresa el punto de partida: ")

destino = input("Ingresa el destino: ")
#origen.lower()
#destino.lower()

# Verificar que existan los lugares

if origen not in G.nodes or destino not in G.nodes:

    print("\nError: el origen o destino no existe en el grafo.")

else:

    # ======================================================
    # RUTA MÁS BARATA
    # ======================================================

    ruta_barata = nx.shortest_path(
        G,
        source=origen,
        target=destino,
        weight="peaje"
    )


    # ======================================================
    # RUTA MÁS RÁPIDA
    # ======================================================

    ruta_rapida = nx.shortest_path(
        G,
        source=origen,
        target=destino,
        weight="tiempo"
    )


    # ======================================================
    # TERCERA RUTA
    # ======================================================

    rutas_posibles = list(
        nx.shortest_simple_paths(
            G,
            origen,
            destino,
            weight="tiempo"
        )
    )

    if len(rutas_posibles) >= 3:
        ruta_alternativa = rutas_posibles[2]

    elif len(rutas_posibles) >= 2:
        ruta_alternativa = rutas_posibles[1]

    else:
        ruta_alternativa = ruta_barata


    # ======================================================
    # MOSTRAR RESULTADOS
    # ======================================================

    mostrar_ruta(
        "RUTA 1 - MÁS BARATA",
        ruta_barata,
        "peaje"
    )

    mostrar_ruta(
        "RUTA 2 - MÁS RÁPIDA",
        ruta_rapida,
        "tiempo"
    )

    mostrar_ruta(
        "RUTA 3 - ALTERNATIVA",
        ruta_alternativa,
        "alternativa"
    )


    # ======================================================
    # DIBUJAR GRAFO
    # ======================================================

    dibujar_grafo(
        ruta_barata,
        ruta_rapida,
        ruta_alternativa
    )
```
